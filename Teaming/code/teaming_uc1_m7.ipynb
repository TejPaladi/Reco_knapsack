```python
import ast
import csv
from statistics import mean

with OUTPUT_FILE.open(newline='', encoding='utf-8') as handle:
    rows = list(csv.DictReader(handle))

volumes = [len(ast.literal_eval(row['team'])) for row in rows]
print({'rows': len(rows), 'volume_mean': mean(volumes), 'volume_min': min(volumes), 'volume_max': max(volumes)})
```
